```python
from ee_solver import *   # loads all solver functions; safe to run in any order
X = [[1, 1],             # Xw = y : each row is one equation, each column one unknown
     [1, -1],
     [1, 0]]
y = [1, 0, 2]            # right-hand side, one number per equation
solve_system(X, y)       # NO bias column is added. picks X^-1 (m = d), left inverse (m > d) or right inverse (m < d)
solution_type(X, y)      # unique / infinitely many / no solution  (rank(X) vs rank([X|y]))
```

## Regression (Lectures 5-6)

```python
from ee_solver import *   # loads all solver functions; safe to run in any order
X = [[-9], [-7], [-5], [1], [5], [9]]   # training features: one row per sample (here 1 feature)
y = [-6, -6, -4, -1, 1, 4]                # training targets: one per sample
r = linreg(X, y,                          # linear regression; the bias column of 1s is added automatically
           lam=0,                         # lam = lambda. 0 -> ordinary least squares, > 0 -> ridge regression
           X_new=[[-1]])                  # new samples to predict (RAW features, no 1s column). None = no prediction
# use bias=False when the question says "no bias / no offset"
# r['w'] = weights (w0 = bias first), r['pred'] = predictions, r['SSE'], r['MSE'] = training errors
```

```python
from ee_solver import *   # loads all solver functions; safe to run in any order
r = linreg(X, y, lam=0.1, X_new=[[-1]])  # ridge: prints PRIMAL (m >= d) or DUAL (m < d) formula automatically
# compare_lambda(X, y, [0, 0.1, 1, 10])  # how w shrinks as lambda grows
```

```python
from ee_solver import *   # loads all solver functions; safe to run in any order
r = polyreg(X, y,        # polynomial regression
           deg=2,          # order of the polynomial (2 -> columns 1, x, x^2)
           lam=0,          # ridge lambda, 0 = none
           X_new=[[-1]])   # RAW new features; they are expanded the same way automatically
# num_params(d=2, deg=3)   # how many parameters: C(d + p, p)
# poly_features([[2, 3]], 2, show=True)   # see the expanded P matrix
```
